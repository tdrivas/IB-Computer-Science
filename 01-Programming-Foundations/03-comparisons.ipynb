<h2> Why do programs need decisions? </h2>

Until now, we have seen that a computer can store data in variables and display information using input and output. However, real programs must do more than just store and print values. They must also compare values, check conditions, and produce results that can later guide decisions.

Examples

- A school system checks whether a student’s grade is above the pass mark.
- A login system checks whether the password entered matches the stored password.
- A shopping website checks whether the total cost is high enough for free shipping.
- A mobile app checks whether a user is online or offline.

In all these situations, the computer performs a comparison and produces a result. That result is often either **True** or **False**.

<h3>What does it mean for a computer to compare?</h3>

Humans compare things naturally.

We tend to say:

- 12 is greater than 5
- A comes before B
- The word "cat" is different from "dog"
- a statement is true
- two answers are not the same

A computer can also compare values, but it does so in a very strict and formal way. It does not “understand” in a human sense. It follows exact rules.

When a computer compares two values, it produces a result. In most cases, that result is one of two possible Boolean values:
- `True`
- `False`

<h3> Comparison Operators </h3>

To compare values, programming languages use comparison operators. These operators instruct the computer what kind of comparison to perform.

Main comparison operators

- Operator `==` means equal to

- Operator `!=` means not equal to

- Operator `>` means greater than

- Operator `<` means less than

- Operator `>=` means greater than or equal to

- Operator `<=` means less than or equal to

These operators are used between two values. hey do not store data. They do not print data. They compare data and generate a **Boolean result**.

<h3> Comparing Characters and Strings </h3>

<h4> How characters are compared </h4>


Computers can also compare characters, such as letters, digits, and symbols. At first, this may seem strange. How can a computer decide whether one letter is “greater” or “less” than another? 

The answer is that characters are stored internally using numeric codes. Each character has an internal representation. So when a computer compares two characters, it is really comparing their numeric codes.

Examples of characters
- A
- B
- a
- 7
- ?


**A character is not magical or abstract for the computer. Internally, it is represented by a number based on ASCII code. So the computer can compare characters because it can compare numbers.**

**Important consequences**
- uppercase letters and lowercase letters are different
- A and a are not the same
- letters can be ordered
- symbols can also be compared


<h4> How strings are compared </h4>

Strings are usually compared **character by character**, from left to right. The computer starts with the first character of each string: 
- If the first characters are different, the comparison is decided immediately.
- If the first characters are equal, it moves to the second characters.
- It continues until it finds a difference.
- If one string ends before the other, length can also matter.

This process is called **lexicographical comparison**, which is similar to dictionary ordering.

<h3> Logical operators </h3>

Sometimes one comparison is not enough. A program may need to combine several conditions. For this, we use logical operators.

Main logical operators:
- not
- and
- or

These operators work with Boolean values and produce Boolean results.





<h4> The and operator </h4>

The operator `and` combines two conditions and returns True only if **both conditions are true**.

Real-life meaning

- A student may be allowed to take the final exam only if attendance is high enough and coursework has been submitted
- A user may access an account only if there is internet connection and the username is correct and the password is correct

Key idea: Both sides must be true. If even one side is false, the whole result becomes false.

```python
7 > 3 and 20 < 4
```

```python
10 == 10 and 5 < 10
```

<h4> The or operator </h4>

The operator `or` combines two conditions and returns True only if **at least one condition is true**.

Real-life meaning

- A student may be allowed into a lab if he/she has a registered member of the course or is an invited guest
- A person has hypertension if diastolic pressure is above 130 or the systolic pressure is above 90. 

Key idea: `or` is more flexible than and. It only needs one true condition.


<h4> The not operator </h4>

The operator `not` reverses a Boolean value:
- `not True` becomes `False`
- `not False` becomes `True`


Real-life meaning

- A system may check whether a user is not logged in.
- A device may trigger an alert if a sensor is not active.
- A website may show a warning if payment is not complete.


Key idea: `not` changes the truth value into its opposite.


<h3> How computers evaluate expressions step by step </h3>


A comparison or logical expression may look simple on the surface, but the computer still evaluates it carefully and systematically.

For example, if an expression contains several parts, the computer:

- evaluates the smaller parts first
- produces intermediate Boolean results
- combines them according to the operators used
- produces one final result

Specifically, the hierarchy of executions is the following:
- arithemtic operations
- comparisons
- logical operations


```python
3 + 2 == 5 and 4 - 5 > 1 # 5 == 5 and -1 > 1 --> True and False --> False
```

<h3> Common beginner misunderstandings </h3>

- Confusing `=` with `==` : Students often use assignment when they really mean comparison.
- Thinking uppercase and lowercase are the same: They are not the same in most comparisons.
- Believing that the computer "understands meaning": It does not compare words by meaning. It compares them by exact symbols and order.
- Forgetting that comparison produces a value: A comparison is not just a question. It is an expression that returns True or False.
- Thinking strings behave like numbers: A number and a string that looks like a number are not the same kind of data. For example, the text "12" is different from the number 12.
